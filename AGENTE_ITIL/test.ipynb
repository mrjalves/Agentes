```python
import pandas as pd
df_tickets = pd.read_csv("df_tickets.csv")

print("✅ df_tickets carregado com sucesso!")
print(f"Linhas: {df_tickets.shape[0]} | Colunas: {df_tickets.shape[1]}")
display(df_tickets.head())
```

```python
# Filtra somente registros em português BR e exporta as colunas solicitadas

df_filtrado = df_tickets[df_tickets["idioma"].astype(str).str.lower().isin(["pt", "pt-br", "pt_br"])].copy()
df_filtrado = df_filtrado.reset_index(drop=True).reset_index().rename(columns={"index": "indexador"})
df_filtrado = df_filtrado[["indexador", "assunto", "descricao", "resposta"]]

arquivo_saida = "tickets_filtrados_ptbr.xlsx"
df_filtrado.to_excel(arquivo_saida, index=False)

print(f"✅ Arquivo salvo com sucesso: {arquivo_saida}")
print(f"Linhas filtradas: {df_filtrado.shape[0]}")
display(df_filtrado.head())
```
